## ⚙️ Algorithm

### Overview
This program follows a simple 3-stage process:

**Stage 1: Build the Character Pool**
→ Ask user which character types they want
→ Combine selected types into one pool string

**Stage 2: Pick Characters Randomly**
→ Loop N times (N = password length the user chose)
→ Each iteration picks one random character from the pool
→ Store each character in a list

**Stage 3: Assemble & Display**
→ Shuffle the list for extra randomness
→ Join the list into a single string
→ Display the final password

### Why We Shuffle After Picking

Imagine user selected only digits and uppercase.

The pool is: `0123456789ABCDEF...Z`

random.choice() always picks from the full pool,
but there is a small chance early picks are all digits
and later picks are all letters — creating a weak pattern.

Shuffling after ensures the final order is
completely unpredictable. ✅

---

### Why We Use a List Instead of a String

❌ String approach (slow):

password = ""

password += "A"    → creates a NEW string every time

password += "3"    → creates another NEW string

password += "@"    → and again...

✅ List approach (efficient):

chars = []

chars.append("A")  → just adds to existing list

chars.append("3")  → no new object created

"".join(chars)     → ONE final string at the end

```python
# import required modules
import string
import random
```

```python
# ========================================
# CELL 5: Build the Character Pool
# ========================================

def char_pool(choices):
    pool = ""

    if choices['upper_case']:
      pool += string.ascii_uppercase

    if choices['lower_case']:
      pool += string.ascii_lowercase

    if choices['digits']:
      pool += string.digits

    if choices['symbols']:
      pool += string.punctuation

    return pool
```

```python
# ========================================
# CELL 6: Generate the Password
# ========================================

def gen_pw(length, pool):
  password_chars = []

  for i in range(length):
    characters = random.choice(pool)
    password_chars.append(characters)

  random.shuffle(password_chars)

  password = "".join(password_chars)

  return password

```

```python
# ========================================
# CELL 7: Get and Validate User Input
# ========================================

def get_user_input():

    # --- Step 1: Get valid password length ---
    while True:
        try:
            length = int(input("Enter password length: "))
            if length < 1:
                print("❌ Length must be at least 1. Try again.")
            else:
                break
        except ValueError:
            print("❌ Invalid input. Please enter a number.")

    # --- Step 2: Get character type preferences ---
    print("\nSelect character types to include:")

    upper_case   = input("Include Uppercase letters? (y/n): ").strip().lower() == "y"
    lower_case   = input("Include Lowercase letters? (y/n): ").strip().lower() == "y"
    digits  = input("Include Digits?           (y/n): ").strip().lower() == "y"
    symbols = input("Include Symbols?          (y/n): ").strip().lower() == "y"

    # --- Step 3: Validate at least one type selected ---
    while not (upper_case or lower_case or digits or symbols):
        print("❌ You must select at least one character type. Try again.\n")
        upper_case   = input("Include Uppercase letters? (y/n): ").strip().lower() == "y"
        lower_case   = input("Include Lowercase letters? (y/n): ").strip().lower() == "y"
        digits  = input("Include Digits?           (y/n): ").strip().lower() == "y"
        symbols = input("Include Symbols?          (y/n): ").strip().lower() == "y"

    return length, upper_case , lower_case, digits, symbols
```

```python
# ========================================
# CELL 8: Main Function
# ========================================

def main():
    print("=" * 40)
    print("       🔐 PASSWORD GENERATOR")
    print("=" * 40)

    while True:

        # --- Step 1: Get user input ---
        length, upper_case , lower_case, digits, symbols = get_user_input()

        # --- Step 2: Build character pool ---
        # Create a dictionary from the boolean choices
        choices = {
            'upper_case': upper_case,
            'lower_case': lower_case,
            'digits': digits,
            'symbols': symbols
        }
        pool = char_pool(choices)

        # --- Step 3: Generate password ---
        password = gen_pw(length, pool)

        # --- Step 4: Display the password ---
        print("\n" + "=" * 40)
        print("✅ Generated Password:")
        print(f"   {password}")
        print("=" * 40)

        # --- Step 5: Ask to generate another ---
        again = input("\nGenerate another password? (y/n): ").strip().lower()

        if again == "y":
            print()
            continue
        else:
            print("\n👋 Thank you for using Password Generator!")
            print("Stay secure! 🔐")
            break
main()        #Run the program
```
